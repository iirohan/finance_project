```python
import networkx as nx
import csv
import matplotlib.pyplot as plt
import pandas as pd
import torch
```

```python
# Supplier Chain Network
year = 2022
postfix = ''
if year == 2022:
  postfix = '_test'

G = nx.read_gml(f"/content/drive/MyDrive/NetworkProject/Dataset/SCNs/NW_{year}_stock_comps_v1_connect{postfix}.gml")
print(f'Nodes:{len(G.nodes)}')
print(f'Edges:{len(G.edges)}')
print(G)

SCNNodes = G.nodes
cc = list(nx.connected_components(G.to_undirected()))
print(f'Total Number of Components:{len(cc)}')
largest_cc = max(cc, key = len)
print(f'Total Number of Nodes in largest Component:{len(largest_cc)}')
```

```python
#Competitor Network
G_Comp = nx.read_gml(f"/content/drive/MyDrive/NetworkProject/Dataset/CN/NW_{year-1}_competition.gml")
print(f'Nodes:{len(G_Comp.nodes)}')
print(f'Edges:{len(G_Comp.edges)}')
print(G_Comp)

CompNodes = G_Comp.nodes
cc_comp = list(nx.connected_components(G_Comp.to_undirected()))
print(f'Total Number of Components:{len(cc_comp)}')
largest_cc_comp = max(cc_comp, key = len)
print(f'Total Number of Nodes in largest Component:{len(largest_cc_comp)}')

largest_subgraph_comp = G_Comp.subgraph(largest_cc_comp)
print(f'Total Number of Edges in largest component:{len(largest_subgraph_comp.edges)}')
```
